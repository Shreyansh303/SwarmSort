### If the session dies

The baseline is one training (roughly 1-1.5 hours), well inside Kaggle's 12-hour limit, so a dead session is simply rerun:
the run is seeded and deterministic, so the rerun trains the same model. Download `results/baseline.json`,
`results/baseline_epochs.csv` (and `results/runs/baseline/` if you want the weights) from the version's Output tab
and commit the two small files to the repository.
